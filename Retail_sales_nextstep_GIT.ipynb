```sql
--Removing rows from tables whose values ​​contain 'NULL'
DELETE FROM dbo.Sell
WHERE CustomerID IS NULL
   OR Gender IS NULL
   OR Age IS NULL
   OR ProductCategory IS NULL
   OR QuantitySold IS NULL
   OR PricePerUnit IS NULL
   OR COGS IS NULL
   OR TotalSaleAmount IS NULL
   OR SaleDate IS NULL
   OR SaleTime IS NULL;
```

```sql
--Checking which category brings the highest profits, broken down by buyer's gender and age
SELECT SUM(TotalSaleAmount) AS 'Sell amount', ProductCategory AS 'Category', Gender AS 'Gender'
FROM dbo.Sell
GROUP BY  Gender, ProductCategory
ORDER BY SUM(TotalSaleAmount) DESC
```

```sql
--Checking which category brings the highest profits and how these profits are distributed in a given category according to the buyer's gender and age
WITH Age AS (
SELECT *,
    CASE
        WHEN Age BETWEEN 10 AND 15 THEN '10-15'
        WHEN Age BETWEEN 15 AND 20 THEN '15-20'
        WHEN Age BETWEEN 20 AND 25 THEN '20-25'
        WHEN Age BETWEEN 25 AND 30 THEN '25-30'
        WHEN Age BETWEEN 30 AND 35 THEN '30-35'
        WHEN Age BETWEEN 35 AND 40 THEN '35-40'
        WHEN Age BETWEEN 40 AND 45 THEN '40-45'
        WHEN Age BETWEEN 45 AND 50 THEN '45-50'
        WHEN Age BETWEEN 50 AND 55 THEN '50-55'
        WHEN Age BETWEEN 55 AND 60 THEN '55-60'
        ELSE 'Pensioners'
    END AS AgeGroup
    FROM dbo.Sell
)
SELECT  ProductCategory AS 'Category', Gender AS 'Gender', AgeGroup, SUM(TotalSaleAmount) - SUM(COGS) AS 'Profit', SUM(TotalSaleAmount) AS 'Sell amount', SUM(COGS) AS 'SumCOGS'
FROM Age
GROUP BY  Gender, ProductCategory, AgeGroup
ORDER BY  Profit DESC




```

```sql
WITH Age AS (
    SELECT *,
        CASE 
            WHEN Age >= 10 AND Age < 15 THEN '10-15'
            WHEN Age >= 15 AND Age < 20 THEN '15-20'
            WHEN Age >= 20 AND Age < 25 THEN '20-25'
            WHEN Age >= 25 AND Age < 30 THEN '25-30'
            WHEN Age >= 30 AND Age < 35 THEN '30-35'
            WHEN Age >= 35 AND Age < 40 THEN '35-40'
            WHEN Age >= 40 AND Age < 45 THEN '40-45'
            WHEN Age >= 45 AND Age < 50 THEN '45-50'
            WHEN Age >= 50 AND Age < 55 THEN '50-55'
            WHEN Age >= 55 AND Age < 60 THEN '55-60'
            ELSE 'Pensioners'
        END AS AgeGroup,
        COGS * QuantitySold AS 'TotalCOGS'
    FROM dbo.Sell
),
Final AS (
    SELECT 
        Gender AS 'Gender',
        AgeGroup AS 'AgeGroup',
        ProductCategory AS 'ProductCategory',       
        SUM(TotalSaleAmount) - SUM(TotalCOGS) AS 'Profit'
    FROM Age
    GROUP BY Gender, AgeGroup, ProductCategory
    
)
SELECT 
    Gender,
    AgeGroup,
    ProductCategory,
    Profit,
    SUM(Profit) OVER() AS 'TotalProfit',
    SUM(Profit) OVER( ORDER BY Profit DESC ) AS 'RunningProfit',
    ROUND(
        100 * SUM(Profit) OVER( ORDER BY Profit DESC) / SUM(Profit) OVER(),
        2
    ) AS 'RunningPercent'

FROM Final
ORDER BY Profit DESC
```
