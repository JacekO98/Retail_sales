**1\. Database Setup**

- **Database Creation**: The project starts by creating a database named p1\_retail\_db.
- **Table Creation**: A table named retail\_sales is created to store the sales data. The table structure includes columns for transaction ID, sale date, sale time, customer ID, gender, age, product category, quantity sold, price per unit, cost of goods sold (COGS), and total sale amount.

```sql
CREATE TABLE retail_sales(
    TransactionID INT,
    SaleDate DATE, 
    Saletime TIME,
    CustomerID INT,
    Gender VARCHAR(6),
    Age INT,
    ProductCategory VARCHAR(100),
    QuantitySold INT,
    PricePerUnit FLOAT,
    COGS FLOAT,
    TotalSaleAmount FLOAT


)
```

```sql
--Write a SQL query to retrieve all columns for sales made on '2022-11-05:
SELECT *
FROM dbo.retail_sales
WHERE SaleDate = '2022-11-05'
```

```sql
--Write a SQL query to retrieve all transactions where the category is 'Clothing' and the quantity sold is more than 4 in the month of Nov-2022:
SELECT *
FROM dbo.retail_sales
WHERE ProductCategory='Clothing' 
AND QuantitySold>'4' 
AND SaleDate BETWEEN '2022-11-01' AND '2022-11-30'

--TUTAJ HINDUS ZROBIŁ INACZEJ I TROCHĘ LEPIEJ. ZOBACZYĆ JAK?
```

```sql
--Write a SQL query to calculate the total sales (total_sale) for each category.:
SELECT DISTINCT(ProductCategory)
FROM dbo.retail_sales

SELECT SUM(TotalSaleAmount) AS 'Całkowita sprzedaż ubrań'
FROM dbo.retail_sales
WHERE ProductCategory = 'Clothing';

SELECT SUM(TotalSaleAmount) AS 'Całkowita sprzedaż elektroniki'
FROM dbo.retail_sales
WHERE ProductCategory = 'Electronics';

SELECT SUM(TotalSaleAmount) AS 'Całkowita sprzedaż kosmetyków'
FROM dbo.retail_sales
WHERE ProductCategory = 'Beauty';


--TUTAJ HINDUS MA INACZEJ ZOBACZYĆ CO I JAK
```

```sql
--Write a SQL query to find the average age of customers who purchased items from the 'Beauty' category.
SELECT ROUND(AVG(Age))
FROM dbo.retail_sales
WHERE ProductCategory='Beauty'
```

```sql
--Write a SQL query to find all transactions where the total_sale is greater than 1000.
SELECT *
FROM dbo.retail_sales
WHERE TotalSaleAmount > '1000'
```

```sql
--Write a SQL query to find the total number of transactions (transaction_id) made by each gender in each category
--to poniżej nie jest moje
SELECT 
    ProductCategory,
    Gender,
    COUNT(*) as total_trans
FROM dbo.retail_sales
GROUP 
    BY 
    ProductCategory,
    Gender
ORDER BY 1
```

```sql
--7.	Write a SQL query to calculate the average sale for each month. Find out best selling month in each year
--Przeanalizować to jeszcze raz bo jest trochę bardziej skomplikowane i interesujące

WITH Miesiace AS (
    SELECT
        FORMAT(SaleDate, 'yyyy') AS Rok,
        FORMAT(SaleDate, 'yyyy-MM') AS RokMiesiac,
        ROUND(AVG(TotalSaleAmount), 2) AS Srednia,
        ROW_NUMBER() OVER (PARTITION BY FORMAT(SaleDate, 'yyyy') 
                           ORDER BY AVG(TotalSaleAmount) DESC) AS rn
    FROM dbo.retail_sales
    GROUP BY FORMAT(SaleDate, 'yyyy'), FORMAT(SaleDate, 'yyyy-MM')
)
SELECT Rok, RokMiesiac, Srednia
FROM Miesiace
WHERE rn = 1
ORDER BY Rok;



```

```sql
--Sprawdzenie jakie lata są zawarte w tabeli
SELECT FORMAT(SaleDate, 'yyyy') AS ROK
FROM dbo.retail_sales
GROUP BY FORMAT(SaleDate, 'yyyy')
```

```sql
--8. Write a SQL query to find the top 5 customers based on the highest total sales 
SELECT TOP 5 CustomerID AS 'Klient', SUM(TotalSaleAmount) AS 'Łączna kwota sprzedaży'
FROM dbo.retail_sales
GROUP BY CustomerID
ORDER BY SUM(TotalSaleAmount) DESC 
```

```sql
--9.	Write a SQL query to find the number of unique customers who purchased items from each category.:
SELECT COUNT(DISTINCT(CustomerID)) AS 'Klient', ProductCategory AS 'Kategoria'
FROM dbo.retail_sales
GROUP BY ProductCategory 
```

```sql
-- 10.	Write a SQL query to create each shift and number of orders (Example Morning <12, Afternoon Between 12 & 17, Evening >17):
WITH NowaTabela AS (
SELECT *,
    CASE
        WHEN SaleTime < '12:00:00' THEN 'Poranek'
        WHEN SaleTime BETWEEN '12:00:00' AND '17:00:00' THEN 'Południe'
        WHEN SaleTime > '17:00:00' THEN 'Wieczór'
    END AS 'Zmiana'
FROM dbo.retail_sales )

SELECT COUNT(QuantitySold), Zmiana
FROM NowaTabela
GROUP BY Zmiana 
```
